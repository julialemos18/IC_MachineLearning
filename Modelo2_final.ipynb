```python
#### padronizar coluna de genero, para que quando aplicar get_dummies(),
### nao seja criada uma coluna para cada formato de entrada
## ex: female / FEMALE / F / woman
```

```python
#padronizar textos, convertendo para minusculas e removendo espaços extra
df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()
```

```python
# Padroniza os valores da coluna Gender
df["Gender"] = df["Gender"].replace({

    # Masculino
    "male": "male",
    "m": "male",
    "mal": "male",
    "male ": "male",
    "make": "male",
    "mail": "male",
    "malr": "male",
    "msle": "male",
    "male-ish": "male",
    "cis male": "male",
    "cis man": "male",
    "guy (-ish) ^_^": "male",
    "ostensibly male, unsure what that really means": "male",
    "male leaning androgynous": "male",

    # Feminino
    "female": "female",
    "f": "female",
    "femail": "female",
    "woman": "female",
    "female (cis)": "female",
    "cis female": "female",
    "cis-female/femme": "female",

    # Trans
    "female (trans)": "trans",
    "trans-female": "trans",
    "trans woman": "trans",

    # Não binário / gênero diverso
    "non-binary": "non-binary",
    "genderqueer": "non-binary",
    "agender": "non-binary",
    "androgyne": "non-binary",
    "fluid": "non-binary",
    "enby": "non-binary",
    "queer": "non-binary",
    "queer/she/they": "non-binary",
    "neuter": "non-binary",

    # Valores inválidos
    "nah": "other"
})
```

```python
df["Gender"].value_counts()
```

```python
#como non-binary, trans e other estao em pouca quantidade, agrupar
df["Gender"] = df["Gender"].replace({
    "trans": "other",
    "non-binary": "other"
})
```

```python
#verifica como ficou
df["Gender"].value_counts()
```

```python
#reconstroi x e y a partir do df ja limpo (com Gender padronizado)
#necessario pois x foi criado antes da padronizacao do Gender

#variaveis preditoras
x = df.drop("treatment", axis=1)

#variavel alvo
y = df["treatment"]

#converte colunas categoricas de x em numericas (Gender ja padronizado)
x = pd.get_dummies(x, drop_first=True)
x = x.astype(float)

#reseta indices para garantir alinhamento entre x e y
x = x.reset_index(drop=True)
y = y.reset_index(drop=True)

print("Formato de x:", x.shape)
print("Formato de y:", y.shape)
```

```python
#padroniza e converte y para 0 e 1
y = y.astype(str).str.strip().str.lower()

y = y.map({
    "no": 0,
    "yes": 1
})

y = y.astype(int)

print(y.value_counts())
```

```python
#confere se sobrou algum valor faltante
#resultado esperado: 0 e 0
print("Valores faltantes em x:", x.isnull().sum().sum())
print("Valores faltantes em y:", y.isnull().sum())
```

```python
#importa o modelo/algoritmo random forest
from sklearn.ensemble import RandomForestClassifier
#importa metricas de avaliacao do modelo
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

#confirmar importacao das bibliotecas
print("sklearn importado com sucesso! :)")
```

```python
#importa o que falta para o split, os modelos e a avaliacao
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

print("bibliotecas importadas com sucesso!")
```

```python
#separa 15% do dataset para validacao final
x_temp, x_validacao, y_temp, y_validacao = train_test_split(
    x,
    y,
    test_size=0.15,
    random_state=42,
    stratify=y
)

#separa o restante em treino e teste
x_train, x_test, y_train, y_test = train_test_split(
    x_temp,
    y_temp,
    test_size=0.20,
    random_state=42,
    stratify=y_temp
)

#mostra o tamanho de cada conjunto
print("Treino:", x_train.shape)
print("Teste:", x_test.shape)
print("Validação:", x_validacao.shape)
```

```python
#cria os modelos que serao comparados
modelos = {
    "Regressão Logística": Pipeline([
        ("scaler", StandardScaler()),
        ("modelo", LogisticRegression(max_iter=5000, random_state=42))
    ]),

    "Random Forest": RandomForestClassifier(random_state=42)
}

#lista para armazenar os resultados de cada modelo
resultados = []

#treina e avalia cada modelo
for nome, modelo in modelos.items():

    modelo.fit(x_train, y_train)

    y_pred_train = modelo.predict(x_train)
    y_pred_test = modelo.predict(x_test)

    acc_train = accuracy_score(y_train, y_pred_train)
    acc_test = accuracy_score(y_test, y_pred_test)
    precision = precision_score(y_test, y_pred_test)
    recall = recall_score(y_test, y_pred_test)
    f1 = f1_score(y_test, y_pred_test)

    resultados.append({
        "Modelo": nome,
        "Acurácia Treino": acc_train,
        "Acurácia Teste": acc_test,
        "Precisão": precision,
        "Recall": recall,
        "F1-score": f1
    })

resultados_df = pd.DataFrame(resultados)
resultados_df
```

```python
#calcula a diferenca entre acuracia de treino e teste
#diferenca grande pode indicar overfitting

resultados_df["Diferença Treino-Teste"] = (
    resultados_df["Acurácia Treino"] - resultados_df["Acurácia Teste"]
)

resultados_df
```

```python
#implementa cross-validation com 5 divisoes
resultados_cv = []

for nome, modelo in modelos.items():
    scores = cross_val_score(
        modelo,
        x_train,
        y_train,
        cv=5,
        scoring="accuracy"
    )

    resultados_cv.append({
        "Modelo": nome,
        "Acurácia média CV": scores.mean(),
        "Desvio padrão CV": scores.std()
    })

resultados_cv_df = pd.DataFrame(resultados_cv)
resultados_cv_df
```

```python
#avalia os modelos no conjunto de validacao final
resultados_validacao = []

for nome, modelo in modelos.items():

    modelo.fit(x_train, y_train)

    y_pred_validacao = modelo.predict(x_validacao)

    acc_validacao = accuracy_score(y_validacao, y_pred_validacao)
    precision_validacao = precision_score(y_validacao, y_pred_validacao)
    recall_validacao = recall_score(y_validacao, y_pred_validacao)
    f1_validacao = f1_score(y_validacao, y_pred_validacao)

    resultados_validacao.append({
        "Modelo": nome,
        "Acurácia Validação": acc_validacao,
        "Precisão Validação": precision_validacao,
        "Recall Validação": recall_validacao,
        "F1-score Validação": f1_validacao
    })

resultados_validacao_df = pd.DataFrame(resultados_validacao)
resultados_validacao_df
```

```python
#GridSearchCV para testar combinacoes de hiperparametros
parametros = {
    "Regressão Logística": {
        "modelo__C": [0.01, 0.1, 1, 10],
        "modelo__solver": ["lbfgs", "liblinear"]
    },

    "Random Forest": {
        "n_estimators": [100, 200],
        "max_depth": [3, 5, 10, None],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4]
    }
}

resultados_tuning = []
melhores_modelos = {}

for nome, modelo in modelos.items():

    grid = GridSearchCV(
        estimator=modelo,
        param_grid=parametros[nome],
        cv=5,
        scoring="accuracy",
        n_jobs=1
    )

    grid.fit(x_train, y_train)

    melhores_modelos[nome] = grid.best_estimator_

    y_pred_treino = grid.best_estimator_.predict(x_train)
    y_pred_teste = grid.best_estimator_.predict(x_test)

    acc_treino = accuracy_score(y_train, y_pred_treino)
    acc_teste = accuracy_score(y_test, y_pred_teste)
    precision_teste = precision_score(y_test, y_pred_teste)
    recall_teste = recall_score(y_test, y_pred_teste)
    f1_teste = f1_score(y_test, y_pred_teste)

    resultados_tuning.append({
        "Modelo": nome,
        "Melhores parâmetros": grid.best_params_,
        "Melhor acurácia CV": grid.best_score_,
        "Acurácia Treino": acc_treino,
        "Acurácia Teste": acc_teste,
        "Precisão Teste": precision_teste,
        "Recall Teste": recall_teste,
        "F1-score Teste": f1_teste
    })

resultados_tuning_df = pd.DataFrame(resultados_tuning)
resultados_tuning_df
```

```python
#avalia os melhores modelos encontrados pelo tuning no conjunto de validacao final
resultados_validacao_tuning = []

for nome, modelo in melhores_modelos.items():

    y_pred_validacao = modelo.predict(x_validacao)

    acc_validacao = accuracy_score(y_validacao, y_pred_validacao)
    precision_validacao = precision_score(y_validacao, y_pred_validacao)
    recall_validacao = recall_score(y_validacao, y_pred_validacao)
    f1_validacao = f1_score(y_validacao, y_pred_validacao)

    resultados_validacao_tuning.append({
        "Modelo": nome,
        "Acurácia Validação após tuning": acc_validacao,
        "Precisão Validação após tuning": precision_validacao,
        "Recall Validação após tuning": recall_validacao,
        "F1-score Validação após tuning": f1_validacao
    })

resultados_validacao_tuning_df = pd.DataFrame(resultados_validacao_tuning)
resultados_validacao_tuning_df
```

```python
#compara a acuracia na validacao antes e depois do tuning
comparacao_tuning = resultados_validacao_df[[
    "Modelo",
    "Acurácia Validação"
]].merge(
    resultados_validacao_tuning_df[[
        "Modelo",
        "Acurácia Validação após tuning"
    ]],
    on="Modelo"
)

comparacao_tuning.set_index("Modelo").plot(kind="bar", figsize=(8, 5))

plt.title("Acurácia na validação antes e depois do tuning")
plt.ylabel("Acurácia")
plt.xlabel("Modelo")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()

comparacao_tuning
```

## Conclusão do Modelo2

- **Regressão Logística:** estável em todos os conjuntos (treino ~83%, teste ~81%, validação ~80%), sem sinais de overfitting.
- **Random Forest:** overfitting real e persistente (treino ~97% vs. validação ~72%), mesmo após o GridSearchCV com `max_depth` limitado. Provável causa: poucas amostras (660 no treino) para 81 variáveis após o one-hot encoding.
- **Comparação com o Modelo1:** lá a acurácia muito alta (~99,8%) levantou suspeita de vazamento de dados / classes facilmente separáveis; aqui o padrão é mais típico de um problema real, e a Regressão Logística generaliza melhor que o Random Forest.
- Modelo mais complexo não significa modelo melhor quando há poucos dados em relação ao número de variáveis.
